### Required Assignment 5.1: Will the Customer Accept the Coupon?

**Context**

Imagine driving through town and a coupon is delivered to your cell phone for a restaurant near where you are driving. Would you accept that coupon and take a short detour to the restaurant? Would you accept the coupon but use it on a subsequent trip? Would you ignore the coupon entirely? What if the coupon was for a bar instead of a restaurant? What about a coffee house? Would you accept a bar coupon with a minor passenger in the car? What about if it was just you and your partner in the car? Would weather impact the rate of acceptance? What about the time of day?

Obviously, proximity to the business is a factor on whether the coupon is delivered to the driver or not, but what are the factors that determine whether a driver accepts the coupon once it is delivered to them? How would you determine whether a driver is likely to accept a coupon?

**Overview**

The goal of this project is to use what you know about visualizations and probability distributions to distinguish between customers who accepted a driving coupon versus those that did not.

**Data**

This data comes to us from the UCI Machine Learning repository and was collected via a survey on Amazon Mechanical Turk. The survey describes different driving scenarios including the destination, current time, weather, passenger, etc., and then ask the person whether he will accept the coupon if he is the driver. Answers that the user will drive there ‘right away’ or ‘later before the coupon expires’ are labeled as ‘Y = 1’ and answers ‘no, I do not want the coupon’ are labeled as ‘Y = 0’.  There are five different types of coupons -- less expensive restaurants (under \$20), coffee houses, carry out & take away, bar, and more expensive restaurants (\$20 - $50).

**Deliverables**

Your final product should be a brief report that highlights the differences between customers who did and did not accept the coupons.  To explore the data you will utilize your knowledge of plotting, statistical summaries, and visualization using Python. You will publish your findings in a public facing github repository as your first portfolio piece.

```python
# Investigate the dataset for missing or problematic data

# A. Initial data-quality investigation


# - data info

print("\033[1m" + "\nData info:\n" + "\033[0m")
data.info()


# - dataset dimensions

print("\033[1m" + "\nDataset size:\n" + "\033[0m")

rows = data.shape[0]
columns = data.shape[1]

print("Rows: " + str(rows) + "\nColumns: " + str(columns))


# - check unique values in categorical columns

categorical_columns = data.select_dtypes(exclude="number").columns

for column in categorical_columns:
    print("\033[1m" + "\n" + column + ":\n" + "\033[0m")
    for value in data[column].unique():
        print("•", value)
```

```python
# - duplicates

print("\033[1m" + "\nDuplicated rows:\n" + "\033[0m")

dup = data.duplicated().sum()

print(
    str(dup)
    + " ("
    + str(round(dup / rows * 100, 2))
    + "%)"
)


# Investigate all observations belonging to duplicated sets

duplicates = data[
    data.duplicated(keep=False)
].sort_values(
    by=list(data.columns)
)

print(
    "\033[1m"
    + "\nRows belonging to duplicated sets:\n"
    + "\033[0m"
)

print("Total:", len(duplicates))

duplicates


# Create a temporary dataset without duplicated observations
# for the remaining data-quality investigation

data_quality = data.drop_duplicates().copy()

rows_quality = data_quality.shape[0]

print(
    "\033[1m"
    + "\nRows after removing duplicates:\n"
    + "\033[0m"
)

print(rows_quality)
```

```python
# - missing counts

print("\033[1m" + "\nMissing data:\n" + "\033[0m")

original_columns = data.columns

missing_data = pd.DataFrame({
    "Total": data_quality[original_columns].isnull().sum(),
    "Percentage": round(
        data_quality[original_columns].isnull().sum()
        / rows_quality * 100, 2
    )
})

print(missing_data)


# - missing percentages - total and by segments

# Automatically identify columns with missing data, excluding 'car'
# because its missingness is assessed separately

cols_with_missing = data_quality.columns[
    data_quality.isnull().any()
    & (data_quality.columns != "car")
]

print(
    "\033[1m"
    + "\nColumns included in missing-data analysis:\n"
    + "\033[0m"
)

for column in cols_with_missing:
    print("•", column)


# Identify rows with at least one missing value
# across the selected columns

data_quality["has_missing"] = (
    data_quality[cols_with_missing]
    .isnull()
    .any(axis=1)
)

print(
    "\033[1m"
    + "\nRows with missing data:\n"
    + "\033[0m"
)

print(
    str(data_quality["has_missing"].sum())
    + " ("
    + str(round(data_quality["has_missing"].mean() * 100, 2))
    + "%)"
)


# Automatically identify variables suitable for segment comparison

segments = [
    column for column in data_quality.columns
    if column not in cols_with_missing
    and column != "car"
    and column != "has_missing"
    and column != "direction_opp"  # redundant with direction_same
    and data_quality[column].nunique() > 1
]


# Check variety of selected segmentation variables

print(
    "\n\033[1m"
    + "Number of unique values by selected segment:"
    + "\033[0m\n"
)

for segment in segments:
    print(
        "•", segment, ":",
        data_quality[segment].nunique()
    )
```

```python
# - calculate missing-data rates for each segment

print("\n\033[1mMissing data by segment\033[0m")

for segment in segments:

    quality = data_quality.groupby(segment)["has_missing"].agg(
        Total="count",
        Missing="sum"
    )

    quality["Missing %"] = round(
        quality["Missing"] / quality["Total"] * 100, 2
    ).astype(str) + "%"

    print("\n\033[1m" + segment + "\033[0m\n")
    print(quality)
```

```python
# - check group sizes for missing-data analysis

# Set the minimum group size used for this diagnostic

min_group_size = 30


# Create lists to collect groups below the minimum group-size criterion

small_total_groups = []
small_complete_groups = []


# Check every group within every segment

for segment in segments:

    quality = data_quality.groupby(segment)["has_missing"].agg(
        Total="count",
        Missing="sum"
    )

    # Calculate observations without missing behavioral data

    quality["Complete"] = (
        quality["Total"] - quality["Missing"]
    )

    # Identify groups with fewer than the minimum number
    # of total observations

    for group in quality.index:

        if quality.loc[group, "Total"] < min_group_size:
            small_total_groups.append(
                (
                    segment,
                    group,
                    quality.loc[group, "Total"],
                    quality.loc[group, "Missing"],
                    quality.loc[group, "Complete"]
                )
            )

        # Separately identify groups with fewer than the minimum
        # number of complete observations

        if quality.loc[group, "Complete"] < min_group_size:
            small_complete_groups.append(
                (
                    segment,
                    group,
                    quality.loc[group, "Total"],
                    quality.loc[group, "Missing"],
                    quality.loc[group, "Complete"]
                )
            )


# Report groups with fewer than the minimum number
# of total observations

print(
    "\n\033[1mGroups with fewer than "
    + str(min_group_size)
    + " total observations:\033[0m"
)

print("Total:", len(small_total_groups))

for segment, group, total, missing, complete in small_total_groups:
    print(
        "•", segment, "-", group,
        "| Total:", total,
        "| Missing:", missing,
        "| Complete:", complete
    )


# Report groups with fewer than the minimum number
# of complete observations

print(
    "\n\033[1mGroups with fewer than "
    + str(min_group_size)
    + " complete observations:\033[0m"
)

print("Total:", len(small_complete_groups))

for segment, group, total, missing, complete in small_complete_groups:
    print(
        "•", segment, "-", group,
        "| Total:", total,
        "| Missing:", missing,
        "| Complete:", complete
    )
```

4. Compare the acceptance rate between drivers who go to a bar more than once a month and are over the age of 25 to the all others.  Is there a difference?


```python
bar_known = bar_coupons.dropna(subset=["Bar"])

condition = (
    bar_known["Bar"].isin(["1~3", "4~8", "gt8"])
    & bar_known["age"].isin(["26", "31", "36", "41", "46", "50plus"])
)

more_1_over_25 = bar_known[condition]
other = bar_known[~condition]

more_1_over_25_acceptance = more_1_over_25["Y"].mean()
other_acceptance = other["Y"].mean()

acceptance_comparison = pd.DataFrame({
    "Group": ["Bar >1/month & age >25", "All others"],
    "Acceptance rate": [more_1_over_25_acceptance, other_acceptance]
})

plt.figure(figsize=(6, 4))

sns.barplot(
    data=acceptance_comparison,
    x="Group",
    y="Acceptance rate",
    hue="Group",
    palette=colors[:2]
)

plt.title("Bar coupon acceptance by customer group")
plt.ylabel("Acceptance rate")
plt.xlabel("")

plt.show()
```

## Observation

Bar coupons were accepted by approximately 70% of respondents over age 25 who went to bars more than once per month, compared with approximately 33% among all other respondents with known bar-visiting frequency. The acceptance rate was therefore about twice as high in the frequent bar-going, over-25 group.
